# **4 Quartis e valores atípicos**

```python
q1 = h.quantile( 0.25 )
q3 = h.quantile( 0.75 )
iqr = q3 - q1

lim_inf = q1 - 1.5 * iqr
lim_sup = q3 + 1.5 * iqr

print( "Q1:", q1, "Q3:", q3, "IQR:", iqr )
print( "Limites:", lim_inf, lim_sup )
```

**q1 e q3**: Encontram o primeiro quartil (25% dos dados) e o terceiro quartil (75% dos dados).

----

**iqr**: Calcula a amplitude interquartil (**Q3 - Q1**), que mede a dispersão dos 50% dados centrais.

----

**lim_inf e lim_sup**: Definem a "cerca" estatística. Qualquer valor abaixo de **lim_inf** ou acima de **lim_sup** é considerado matematicamente um outlier (**como o aluno com 20 horas de estudo, que ultrapassaria o limite superio**r).

----

```python
atipicos = df[ ( h < lim_inf ) | ( h > lim_sup )]
atipicos
```

**(h < lim_inf)**: Cria uma condição (**máscara booleana**) para verificar quais valores na coluna **h** (**horas**) são menores que o limite inferior.

---

**|**: É o operador lógico "OU" (**OR bitwis**e) no Pandas. Ele une a condição inferior com a superior.

----

(**h > lim_sup**): Cria a condição para verificar quais valores são maiores que o limite superior (**que é onde o aluno com 20 horas se encaixa**).

----

**df[...]**: Aplica essa regra de seleção no DataFrame original df, mantendo apenas as linhas onde a condição é verdadeira.

----

**atipicos**: Armazena esse novo DataFrame filtrado e, na última linha, exibe o resultado na tela.

----

```python
meses = list( range(1, 13))
vendas_mes = pd.Series( [120, 135, 150, 142, 160, 175, 170, 190, 205, 198, 220, 240], index = meses)

media_movel = vendas_mes .rolling( window = 3 ).mean()

plt.plot( vendas_mes, marker = "o", label = "vendas")
plt.plot( media_movel, linestyle = "--", label = "media movel (3 meses)")
plt.title( "Vebas ao longo do ano")
plt.xlabel( "Meses")
plt.ylabel( "Vendas")
plt.legend()
plt.show()
```

**meses = list(range(1, 13))**: Cria uma lista com números de 1 a 12, representando os meses do ano (**janeiro a dezembro**).

----

**vendas_mes = pd.Series([...], index=meses**): Transforma os valores de vendas em uma Série do Pandas, associando cada valor de venda ao seu respectivo mês (**índice**).

----

**.rolling(3)**: Define uma janela deslizante (**rolling window**) de 3 períodos consecutivos (**agrupando os meses de 3 em 3**).

----

**.mean()**: Calcula a média aritmética dos valores dentro dessa janela de 3 meses. Isso serve para suavizar as flutuações bruscas e destacar a tendência geral de crescimento ou queda ao longo do tempo.

----

**plt.plot(vendas_mes, marker="o", label="vendas")**: Plota a linha com as vendas reais, colocando um marcador circular (**marker="o"**) em cada ponto mensal.

----

**plt.plot(media_movel, linestyle="--", label="media movel (3 meses)")**: Plota a linha da média móvel utilizando um estilo tracejado (**linestyle="--"**).

----

# **8 Desafio: organize e explore sozinho**

```python
turma = pd.DataFrame({
    "Aluno": ["Ana", "Bruno", "Carla"],
    "P1": [7.0, 5.5, 9.0],
    "P2": [8.0, None, 8.5],
    "P3": [None, 6.0, 7.5],
    "Curso/Turno": ["Estatistica-Noite", "Ciencia de Dados-Manha", "Estatistica-Noite"]
})
turma
```

1. separa Curso/Turno em duas colunas, curso e turno;
2. transforma as colunas P1, P2 e P3 em duas colunas, prova e nota (formato longo);
3. decide e registra o que fazer com as provas não realizadas;
4. calcula a nota média de cada prova e a nota média de cada aluno;
5. cria um gráco de barras com a nota média de cada prova

**Passo 1**: Separar Curso/Turno em duas colunas (curso e turno)

```python
turma[["curso", "turno"]] = turma["Curso/Turno"].str.split( "-", expand=True )
turma = turma.drop( columns="Curso/Turno" )
```

**Passo 2**: Transformar as colunas de provas em formato longo (melt)

```python
turma_long = turma.melt(
    id_vars=["Aluno", "curso", "turno"],
    value_vars=["P1", "P2", "P3"],
    var_name="prova",
    value_name="nota"
)
```

**Passo 3**: Tratar as provas não realizadas (valores ausentes / NaN)

Decisão: Preencher os valores ausentes com 0 (zero).

Justificativa: Se o aluno não realizou a prova (como o Bruno na P2 e a Ana na P3), considerar como nota zero reflete o impacto real na média geral e evita distorções ou perda de registros.

```python
turma_long["nota"] = turma_long["nota"].fillna( 0 )
```

**Passo 4:** Calcular a nota média de cada prova e de cada aluno

```python
# Média de cada prova
media_prova = turma_long.groupby( "prova" )["nota"].mean()
print( "Média por Prova:" )
print( media_prova )

# Média de cada aluno
media_aluno = turma_long.groupby( "Aluno" )["nota"].mean()
print( "\nMédia por Aluno:" )
print( media_aluno )
```

**Passo 5**: Criar um gráfico de barras com a nota média de cada prova

```python
import matplotlib.pyplot as plt

media_prova.plot( kind="bar", color="skyblue", edgecolor="black" )
plt.title( "Nota Média por Prova" )
plt.xlabel( "Prova" )
plt.ylabel( "Nota Média" )
plt.xticks( rotation = 0 )
plt.ylim( 0, 10 )
plt.grid( axis="y", linestyle="--", alpha=0.7 )
plt.show()
```

**Pergunta 17**. Qual foi a sua decisão sobre as provas não realizadas e por quê? Como essa decisão
altera a média de cada prova?

**Resposta:** Decisão e justificativa: A decisão foi preencher os valores ausentes (NaN) com 0 (zero). Justificativa: se o aluno não realizou a prova, considerar zero reflete o impacto real na sua pontuação sem precisar excluir o registro do aluno ou alterar a estrutura da tabela.

Como a decisão altera a média de cada prova: Essa decisão reduz (puxa para baixo) a média da prova. Como o zero entra no somatório das notas enquanto a quantidade total de alunos (divisor) permanece a mesma, a inclusão de uma nota nula diminui o valor final da média daquela avaliação.

**Pergunta 18**. Qual a granularidade da base final? Cada linha representa o quê?

**Resposta:** A granularidade da base final (turma_long) é um registro por aluno e por prova. Cada linha representa a nota obtida por um aluno específico em uma avaliação específica (seja a P1, P2 ou P3).

**Pergunta 20**. Cite duas situações em que a organização inadequada da base poderia levar a uma
conclusão errada.

**Resposta:** Duas situações em que a organização inadequada da base pode levar a conclusões erradas são:

Tratamento incorreto de valores ausentes (missing values): Se valores nulos (como a ausência de compras em um mês) forem simplesmente ignorados ou removidos em vez de tratados adequadamente (como preenchidos com zero), as métricas estatísticas — como a média de vendas por cliente ou o faturamento total por período — serão distorcidas, superestimando o desempenho real.

Falta de normalização ou duplicação em cruzamentos (merges): Se os dados estiverem desestruturados e duplicados na mesma tabela, uma operação de agrupamento (groupby) somará valores repetidos indevidamente, gerando totais inflados e falsos (por exemplo, duplicando vendas de um estado), o que levaria a uma decisão comercial completamente equivocada baseada em números irreais.
